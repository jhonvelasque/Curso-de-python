# Joins
- Unión con llave en común
- Merge con llave distinta
- Unión con múltiples llaves
- Left / Rigth / Inner / Outer Join

```python
df_base.FUENTE.value_counts()
```

```python
df_base.groupby('FUENTE')['CODIGO'].count()
```

```python
# Creando DataFrame Descriptivo
df_desfuente = pd.DataFrame({'FUENTE':['API','WBS'],
                             'FNT':['API','WBS'],
                             'DESFUENTE':['API de Google','Webscrapimng a Google Maps'],
                             'ORIGEN':['UM_COORD_EMPRESA','WEB PAGE']
                            })
df_desfuente.head()
```

```python
# Cruce utilizando un campo en común
df_cruce = pd.merge(df_base,
                    df_desfuente[['FUENTE', 'DESFUENTE']],
                    on = 'FUENTE',
                    how = 'left' # right, inner
                    )
df_cruce.head(2)
```

```python
# Cruce utilizando diferentes campos
df_cruce = pd.merge(df_base,
                    df_desfuente[['FNT', 'DESFUENTE']],
                    left_on = ['FUENTE'],
                    right_on = ['FNT'],
                    how = 'inner'
                    )
df_cruce.head(2)
```

```python
# BONUS: Borrando una columna en un dataframe: .drop()
df_cruce.drop('FNT', inplace=True, axis=1)
df_cruce.drop(['X', 'Y'], inplace=True, axis=1)
df_cruce.head(2)
```

```python
# BONUS: Borrando una columna en un dataframe: del
del df_cruce['PROV']
df_cruce.head(2)
```

```python
# BONUS: Borrando una columna en un dataframe: .pop()
df_cruce.pop('DPTO')
```

```python
dict_clase = {'codigo':[9876654,7665498,6598764,5498766,8976654],
             'edad':[26,np.nan,30,42,38],
             'ingreso':[2500,2500,3500,4800,4100],
             'estado_civil':['Soltero','Soltero','Casado','Casado','Soltero'],
             'fecmatricula':['25/07/2022','21/07/2022','28/07/2022','26/07/2022','22/07/2022']
            }
df_clase = pd.DataFrame(dict_clase)
df_clase
```

# Cambiando tipo de dato

```python
# Campos Calculados: Cambiando tipo de dato: .astype()
df_clase['DNI'] = df_clase['codigo'].astype('str').str.zfill(8)
df_clase.head(2)
```

```python
df_clase['EC'] = df_clase['estado_civil'].astype('str').str[0].str.upper()
df_clase.head(2)
```

```python
df_clase['next_year'] = df_clase['fecmatricula'].astype('str').str[-4:].astype('int')+1
df_clase.head(2)
```

```python
df_clase['month'] = df_clase['fecmatricula'].astype('datetime64[ns]').dt.month
df_clase.head(2)
```

```python
df_clase['day'] = df_clase['fecmatricula'].astype('datetime64[ns]').dt.day
df_clase.head(2)
```

```python
df_clase['cod_mes'] = df_clase['fecmatricula'].astype('datetime64[ns]').dt.strftime('%Y%m')
df_clase.head(2)
```

```python
df_clase['formatted_date'] = df_clase['fecmatricula'].astype('datetime64[ns]').dt.strftime('%d de %m del %Y')
df_clase.head(2)
```

```python
df_clase['cod_date'] = pd.to_datetime(df_clase['formatted_date'], format='%d de %m del %Y').dt.strftime('%Y%m%d')
df_clase.head(2)
```

# Lambda Functions en DataFrames

```python
df_clase = df_clase[['codigo','edad','ingreso','estado_civil','fecmatricula', 'DNI']]
df_clase.head(2)
```

```python
# Aplicando Lambda Functions en una columna
# # Cuando se utilizan series no es necesarip asignar un axis
df_clase['TIP_INGRESO'] =  df_clase['ingreso'].apply(lambda x: 'ALTO' if x > 3000 else 'BAJO')
df_clase.head(3)
```

```python
df_clase['CODMESMATRICULA'] = df_clase['fecmatricula'].apply(lambda x: x.split('/')[2]+x.split('/')[1])
df_clase.head(1)
```

```python
# Aplicando lambda function a multiples columnas
# Al utilizar Dataframe, es necesario utilizar un axis
df_clase['TIP_INGRESO2'] = df_clase[[
    'ingreso', 'edad'
    ]].apply(lambda x:
             'ALTO' if x['ingreso'] > 4000 else 'MEDIO' if x['ingreso'] > 2000 and x['edad'] < 30 else 'BAJO',
             axis = 1)  # (Manda la columna) Aplica funcion a cada fila
df_clase.head(2)
```

```python
# Utilizando lambda function con una funcion ya definida (Casos complejos)
def genera_tipingreso(ingreso, edad):
  if ingreso > 4000:
    result = 'ALTO'
  elif ingreso > 2000 and edad < 30:
    result = 'MEDIO'
  else:
    result = 'BAJO'
  return result
```

```python
df_clase['TIP_INGRESO3'] = df_clase[[
    'ingreso', 'edad']].apply(lambda x: genera_tipingreso(x['ingreso'], x['edad']), axis=1)
df_clase.head(2)
```
