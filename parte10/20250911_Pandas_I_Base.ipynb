- Bono: Renombrado de columnas
- Bono: Reseteo de índices

# Agrupaciones Directas

```python
df_base = df_drive_pkmn[:16]
df_base
```

```python
df_base.head(1)
```

```python
# Método groupby: Permite agrupar grandes cantidades de datos y calcular operaciones en estos grupos
gb_base = df_base.groupby('Type 1')
type(gb_base)
```

```python
# Cantidad de registros agrupados
# # Agrupa y muestra la cantidad de registros con un valor
gb_base = df_base.groupby('Type 1')
gb_base.count()
```

```python
gb_base = df_base.groupby('Type 2', dropna=False)
gb_base.count()
```

```python
# # Agrupación mostrando una Serie
gb_base = df_base.groupby('Type 1')['#']
gb_base.count()
```

```python
# # Agrupación mostrando un Dataframe
gb_base = df_base.groupby('Type 1')[['#', 'Type 2']]
gb_base.count()
```

```python
# Máximo valor de una agrupación
gb_base = df_base.groupby('Type 1')[['Attack', 'Sp. Atk']]
gb_base.max()
```

```python
# Mínimo valor de una agrupación
gb_base = df_base.groupby('Type 1')[['Attack', 'Sp. Atk']]
gb_base.min()
```

```python
# Mediana de una agrupación
gb_base = df_base.groupby('Type 1')[['Attack', 'Sp. Atk']]
gb_base.median()
```

```python
# Promedio de una agrupación
gb_base = df_base.groupby('Type 1')[['Attack', 'Sp. Atk']]
gb_base.mean()
```

# Agrupaciones por agregación

```python
# Total de registros, mínimo y máximo valor por determinada columna
gb_base = df_base.groupby('Type 1')[['Attack']]
gb_base.agg(['count','min','max'])
```

# Agrupaciones con múltiples campos

```python
# Total de registros, mínimo y máximo valor por determinada columna
gb_base = df_base.groupby('Type 1')[['Attack', 'Sp. Atk']]
gb_base.agg(['count','min','max'])
```

# Agrupaciones independientes al campo

```python
# Filtro de tipo de agrupación por campo
gb_base = df_base.groupby('Type 1')[['#', 'Attack', 'Sp. Atk']]
gb_base.agg({'#':'count','Attack':['min','max'],'Sp. Atk':'max'})
```

# Agrupacion múltiple

```python
gb_base = df_base.groupby(['Type 1','Type 2'], dropna=False)[['Attack', 'Sp. Atk']]
gb_base = gb_base.agg({'Attack':'max','Sp. Atk':['min','max']})
gb_base
```

```python
print(type(gb_base))
```

```python
print('Total COLUMNAS:',len(gb_base.columns))
print('COLUMNAS:',gb_base.columns)
```

```python
print('Total INDICES:',len(gb_base.index))
print('INDICES:',gb_base.index)
```

# Bono: Renombrado de columnas

```python
# Renombrado de todas las columnas de un Dataframe
gb_base.columns = ['max_attack', 'min_sp_attack', 'max_sp_attack']
gb_base
```

```python
# Renombrado de determinadas columnas en un DataFrame
gb_base.rename(columns={'min_sp_attack': 'min_spatk', 'max_sp_attack': 'max_spatk'}, inplace=True)
gb_base
```

# Bono: Reseteo de índices

```python
# Reseteando índices en un Dataframe
gb_base.reset_index(inplace=True)
gb_base
```

```python
# Reseteo de índice sin guardar el índice actual
gb_base.reset_index(drop=True)
```
